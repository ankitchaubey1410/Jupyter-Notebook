## EXECEPTION HANDLING

###### There are two stages where error may appear in a program
###### 1.During compilation - Syntax error
###### 2.During execution - Exceptions

#### Syntax Error
###### 1.Something is not written according to the program grammer.
###### 2.This error is detected by the compiler / interpreter.
###### 3.You can solve it by correcting the program.

```python
# Example of syntax error
print 'ichigo'
```

```python
# Corrected code
print('ichigo')
```

#### Other examples of syntax error
###### 1.misspelling a keyword
###### 2.not using indentation
###### 3.empty for, loop, funtion, if else
###### 4.leaving symbols like colon, comma or brackets

```python
a = 4
for i in range(a)
    print(i)
```

```python
a = 6
iff a == 4:
    print(a)
```

```python
# This is a syntax error but python identify it as indentation error
a = 5
for i in range(a):
print(a)
```
