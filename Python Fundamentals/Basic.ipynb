## Type Conversion

```python
# Implicit vs Explicit
# Example of Implicit
print(5.6+4)
print(type(5.6),type(4))
# Example of Explicit
print(9.9+int('10'))
print(type(9.9),type('10'))
```

## Literals

```python
# Literals are the values given to the variables
a = 0b1010 # Binary Literals
b = 100 # Decimal Literals
c = 0o310 # Octal Literals
d = 0x12c # Hexadecimal Literals

# Float Literal
float_1 = 10.5
float_2 = 1.5e2 # 1.5 * 10^2
float_3 = 1.5e-3 # 1.5 * 10^-3

# Complex Literal
x = 3.14j

print(a,b,c,d)
print(float_1,float_2,float_3)
print(x,x.imag,x.real)
```
